# Loops

Let's say that I have a string, `'abcd'`, and I want to print each of the characters in this string.

```python
s = 'abcd'

print(s[0])  # index 0 is the first character
print(s[1])
print(s[2])
print(s[3])
```

Technically speaking, the above is totally fine Python.

But ... it feels weird and bad, and we should really try to change.

The principle that can help us here is **DRY**, "don't repeat yourself."

As you can see, my code in the previous cell basically repeated itself on each line. Only the index was changing from lines 3-6. Maybe we could just say to Python, "`print(s[n])`, where `n` starts at 0 and goes up through 3."

But in Python, we can do it even easier than that: We can just say, "Give me each of the characters in s, from the start through the end."

```python
# for loop

s = 'abcd'

print('Before')
for one_character in s:
    print(one_character)
print('After')    
```

# Syntax of a `for` loop

1. We start with the keyword `for`.
2. Then we have the *loop variable*, in this case, `one_character`. This is the variable that will get a new value with each iteration.
3. Then we haev the keyword `in`.
4. Then we have the value -- the *iterable* as it is known -- over which we'll be iterating. In this case, it's a string, and we'll get each character from the string.
5. At the end of the line, we have a `:` -- because after that, we have an indented loop body, a Python block.
6. The loop body is indented, and can contain ANY PYTHON CODE YOU WANT. Typically, you want to keep it short and clear.
7. The loop will execute once for each item that the iterator provides (`s`, in this case). After that, execution of the program continues.

# Think of it as a conversation

1. `for` turns to the iterable value (`s`) and asks: Are you really iterable?
    - If not, then the loop exits with an error.
2. `for` says: If you're iterable, then give me your next value.
    - If there are no more values, then the loop exits.
3. If there is a value, then it is assigned to the loop variable (`one_character`).
4. The loop body executes with variable assigned.
5. After the loop body executes, we go back to step 2, asking for the next value.

# Some myths/things to explain

1. I like to call my loop variable `one_character` or `one_digit` or `one_THING`. This is for me! This has no effect whatsoever on what type of value I get with each iteration. Nor does it affect how many values I'll get.
2. In a Python `for` loop, as we just discussed, the loop doesn't know when it'll stop. It's up to the iterable value to know how many times to iterate, what values to provide with each iteration, and when it'll stop.

# Exercise: Vowels, digits, and others

1. Define three variables -- `vowels`, `digits`, and `others`, and assign them all to be 0.
2. Ask the user to enter some text, and assign to `text`.
3. Go through the text, one character at a time, in a `for` loop:
    - If the character is a vowel (a, e, i, o, u), then add 1 to `vowels`
    - If the character is a digit (0-9), then add 1 to `digits`
    - In other cases, add 1 to `others`
4. Print each of the variables.

Example:

    Text: hello!! 123
    vowels: 2
    digits: 3
    others: 6

```python
vowels = 0
digits = 0
others = 0

text = input('Text: ').strip()

for one_character in text:
    if one_character in 'aeiou':     # is it a vowel?
        print(f'\t{one_character} is a vowel')
        vowels += 1                  #   add 1 to vowels
    elif one_character.isdecimal():  # is it a number 0-9?
        print(f'\t{one_character} is a digit')
        digits += 1                  #   add 1 to digits
    else:
        print(f'\t{one_character} is other')
        others += 1                  # otherwise, add 1 to others

print(f'vowels = {vowels}')        
print(f'digits = {digits}')
print(f'others = {others}')
```

# What about looping a number of times?

```python
# I'm so excited to be teachign Python again!

print('Hooray!')
print('Hooray!')
print('Hooray!')  # doesn't this violate the DRY rule?
```

```python
# let's use a loop!

for counter in 3:      # for turns to 3, and asks: Are you iterable? 
    print('Hooray!')
```

In order to iterate a number of times, you need to use the `range` function. `range(3)` is designed to be used in a loop to iterate 3 times!

```python
for counter in range(3):     
    print('Hooray!')
```

Didn't we say before that when `for` is iterating over a loop, it asks the iterator each time to give it the next value?

What values, if any, are being put in `counter`?



```python
for counter in range(3):     
    print(f'{counter} Hooray!')
```

# Let's talk about `range`

If you want to iterate `n` times, you should use a `for` loop and `range(n)`. This will give you `n` iterations.

The value assigned to the loop variable with each iteration will be an integer, starting with 0 and ending with `n-1`.
