# Loops

Let's say that I have a string, `'abcd'`, and I want to print each of the characters in this string.

```python
s = 'abcd'

print(s[0])  # index 0 is the first character
print(s[1])
print(s[2])
print(s[3])
```

Technically speaking, the above is totally fine Python.

But ... it feels weird and bad, and we should really try to change.

The principle that can help us here is **DRY**, "don't repeat yourself."
