# Loops

Let's say that I have a string, `'abcd'`, and I want to print each of the characters in this string.

````python
s = 'abcd'

print(s[0])  # index 0 is the first character
print(s[1])
print(s[2])
print(s[3])
````

Technically speaking, the above is totally fine Python.

But ... it feels weird and bad, and we should really try to change.

The principle that can help us here is **DRY**, "don't repeat yourself."

As you can see, my code in the previous cell basically repeated itself on each line. Only the index was changing from lines 3-6. Maybe we could just say to Python, "`print(s[n])`, where `n` starts at 0 and goes up through 3."

But in Python, we can do it even easier than that: We can just say, "Give me each of the characters in s, from the start through the end."

````python
# for loop

s = 'abcd'

print('Before')
for one_character in s:
    print(one_character)
print('After')    
````

# Syntax of a `for` loop

1. We start with the keyword `for`.
2. Then we have the *loop variable*, in this case, `one_character`. This is the variable that will get a new value with each iteration.
3. Then we haev the keyword `in`.
4. Then we have the value -- the *iterable* as it is known -- over which we'll be iterating. In this case, it's a string, and we'll get each character from the string.
5. At the end of the line, we have a `:` -- because after that, we have an indented loop body, a Python block.
6. The loop body is indented, and can contain ANY PYTHON CODE YOU WANT. Typically, you want to keep it short and clear.
7. The loop will execute once for each item that the iterator provides (`s`, in this case). After that, execution of the program continues.

# Think of it as a conversation

1. `for` turns to the iterable value (`s`) and asks: Are you really iterable?
    - If not, then the loop exits with an error.
2. `for` says: If you're iterable, then give me your next value.
    - If there are no more values, then the loop exits.
3. If there is a value, then it is assigned to the loop variable (`one_character`).
4. The loop body executes with variable assigned.
5. After the loop body executes, we go back to step 2, asking for the next value.

# Some myths/things to explain

1. I like to call my loop variable `one_character` or `one_digit` or `one_THING`. This is for me! This has no effect whatsoever on what type of value I get with each iteration. Nor does it affect how many values I'll get.
2. In a Python `for` loop, as we just discussed, the loop doesn't know when it'll stop. It's up to the iterable value to know how many times to iterate, what values to provide with each iteration, and when it'll stop.

# What about the index?

In other languages, we can only iterate with an integer/index, and use that to retrieve the characters.

In Python, we don't need that, because we get the characters themselves. 

However, we do sometimes want to print the index, or to use it.

It might be nice to print each of the characters in a string alongside its index.

There are at least two ways to do this.

````python
# option 1: manually

text = 'abcde'
index = 0

for one_character in text:
    print(f'{index} {one_character}')
    index += 1

# here, we're using the for loop to calculate the index ourselves, and add to it with each iteration    
````

````python
# option 2: automatically, with "enumerate"
# enumerate is a built-in function that is designed exactly for this
# it is invoked on an iterable value, and then returns TWO values with each iteration
# so when we use enumerate, we're not iterating over text -- we're iterating over enumerate(text)

text = 'abcde'

for index, one_character in enumerate(text):   # we get two values, and thus have TWO LOOP VARIABLES!
    print(f'{index} {one_character}')
````

# Controlling our loops

It's great that we can iterate over all of the characters in a string. However, what if we have accomplished our goal halfway through the loop? Do we need to keep iterating? What if we realize very quickly that the current iteration isn't needed or useful, and that we should just move onto the next iteration?

These are both very common thing in `for` loops. There are two special statements in Python to handle these:

- `break` -- this means: Exit the loop right now. Right away. (And then continue with the rest of the program.)
- `continue` -- this means: Exit the *current iteration* right now, going onto the next iteration (unless there aren't any more)

I typically use `break` if I have accomplished my goal. I typically use `continue` if I know that I don't need the current iteration.

````python
# continue 

text = 'abcde'
to_ignore = 'd'

for one_character in text:
    if one_character == to_ignore:
        continue   # go onto the next iteration

    print(one_character)
````

````python
# break 

text = 'abcde'
to_ignore = 'd'

print('Begin')
for one_character in text:
    if one_character == to_ignore:
        break  # stop the loop

    print(one_character)
print('End')    
````

It's very commmon to have `if` statemnts at the top of a loop body. Those `if` statements check if we have fulfilled the requirements for either `break` or `continue`. The rest of the loop body can then assume, without even an `else` block, that we don't need to worry about `break` or `continue`.

# `while` loops

So far, we've been talking about `for` loops, which are great when you know how many times you want to iterate: Use `range` for a number. Or iterate over every character in a string.

But if you don't know how many times you'll want to iterate? You know when you want to stop, but you don't know how many times that'll take.

That's where a `while` loop comes in: It is like an `if` statement, where it looks to its right and expects a `True` or `False` value. If the value is `True`, then it'll run the loop body and then go back to the `while` loop to check the condition again. If the value is `False`, then the `while` loop doesn't run, and goes onto the next part of the program.  It's like an `if` that runs repeatedly, so long as the condition is `True`.



````python
x = 5

while x > 0:   # so long as this is true...
    print(x) 
    x -= 1     # reduce x by 1
````

````python
# classic (if a bit scary-looking) piece of Python code:

while True:   # yes, an infinite loop!!
    name = input('Enter your name: ').strip()

    if name == '':   # break out of the loop if we got an empty string...
        break

    print(f'Hello, {name}')
````

# Lists

A list is one of the most important data structures in Python. It's our "default container" type, meaning that if you have a bunch of values and you want to store them, you're likely to use a list.

A Python list feels/seems similar to what some other languages call an "array." There are technical differences, but Python people call them "lists." Some of the reasons why we don't call them arrays is that Python lists (a) can have elements of different types and (b) Python lists can change in size after they're created. 

## List facts:

- You create a Python list with `[]`.
- The elements of the list can be *any* Python values whatsoever, without any restrictions.
- The elements are separated from one another with commas.
- Even though this is not enforced by the Python language, it's traditional for a list to contain only elements of a single type. You can put whatever you want in a list, but you normally want a list of strings, a list of integers, a list of lists...

## When would you use a list?

- List of files in a directory
- List of employees at a company
- List of cars in the company's fleet
- List of packages that the delivery van needs to drop off

# Strings are immutable, and lists are mutable

Once you've created a string, you cannot change it. You can assign a new string to the same variable, but the string you created and assigned is permanent. If it contains `'abcde'`, then it will always contain those letters; you want a longer or shorter string, you must create a new one based on the existing one.

Strings are *immutable*, they cannot be changed.

By contrast, lists are *mutable*, they can be changed! What does this mean?

- You can replace an existing element with another one
- You can add new elements, making the list longer
- You can remove existing elements, making the list shorter
