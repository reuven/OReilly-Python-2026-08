# What are modules?

Twice so far, I've talked to you about the DRY rule -- don't repeat yourself! 

- If the same line of code is used several times in a row, then we can replace those lines with a loop.
- If the same code is used in several places in your program, then you can use a function instead.

What happens if you have the same code in several different programs?

Examples:

- Menuing system
- Login system
- Secret formula that you don't want to have in multiple places

In all of these cases, you can't just define a function and use it, because you have multiple programs.

So the solution in most programming languages is a *library*, a collection of data and functionality that you can use repeatedly, and in multiple programs.

In Python, we call a library a "module" or a "package."

- A module is one Python file that we can "import" into our program and use
- A package is a directory/folder of modules, a collection that we can pick and choose from

We use modules all of the time in Python. But in Python, they do more than that:

- Modules provide us with data and functions that we'll use in multiple programs
- Modules provide us with *namespaces*, like surnames for variables
- Packages allow us to distribute a module (or collection of modules) to other Python programmers, in our own organization or even elsewhere.

# Exercise: Guessing game

1. Generate a random `secret` number by using the `randint` function in the `random` module, choosing one from 0-100.
2. Ask the user, repeatedly, to guess the number:
    - If they guess correctly, say "You got it!" and exit from the program
    - If it's too low, say "Too low!"
    - If it's too high, say "Too high!"
3. If they got it wrong, have them try again
4. Print the number of guesses it took for them to get it right.

````python
import random

secret = random.randint(0, 100)
guess_counter = 0

while True:
    guess_text = input('Guess: ').strip()

    if not guess_text.isdecimal():
        print(f'{guess_text} is not numeric; try again')
        continue
    
    guess = int(guess_text)
    guess_counter += 1
    
    if guess == secret:
        print(f'You got it in {guess_counter} guesses!')
        break
    elif guess < secret:
        print(f'Too low!')
    else:
        print(f'Too high!')
````

# Two more ways to import

I just told you that I can import Pandas with `import pandas`.

No one does that. Everyone, but everyone, uses an alias for `pandas`, `pd`.

How can you create the alias? With the `as` option to `import`.

````python
import pandas as pd   # this imports pandas.py, but then assigns it to the variable pd, and not to pandas
````

Similarly, you can say

```python

```

This imports `MODULE` into memory, but only defines one name, `ALIAS`, which is assigned to `MODULE.NAME`.

# The 4 (5?) different versions of `import`

- `import MODNAME`
- `import MODNAME as ALIAS`
- `from MODNAME import NAME`
- `from MODNAME import NAME as ALIAS`

There is a 5th version of this:

`from MODNAME import *`

This means: 

- Load `MODNAME` into memory
- Don't define it as a variable
- Instead, iterate over every name in the module, and define it as a global variable

In other words, if a module defines `x`, `y`, and `z`, we'll have those three variables defined. Not as attributes on the module, but as global variables.

# What really happens when we say `import`?

Consider this:

- Before we `import menu`, the `menu.menu` function doesn't exist.
- After we `import menu`, it does.
- Sometime during the import, the function was defined.
- This means that `def menu` inside of `menu.py` executed when we imported the menu.

This is something that's very surprising to a lot of people: When you say `import menu`, you aren't just loading it into memory. Rather, you are executing the entire module, from start to finish.

If a module is very big, very slow, very un-modular, this can take some time.

This means that the module is executed, and functions are only around before `def` executes, and variables are only around because `=` executes.

This raises an interesting (if slightly evil) question: What if we put executable code inside of our module?

````python
import mymod
````

But the fact that the module is executed when it is imported has lots of implications, including what we can execute.

# `__name__` is magical

Everywhere, in every Python program, the variable `__name__` is defined. It will always contain a string value. What value it contains depends on the context:

- In the first file that our program executes, `__name__` contains the special string `'__main__'`. (Don't mix them up; the first is a variable, and the second is a string.)
- Everywhere else, including in every module that was imported, `__name__` contains the name of the module.

This allows us to distinguish between when a program has been imported as a module, and when it's being run standalone. 

In other words:

- If we run `mymod.py` from the command line, then `__name__` will be (as we saw) `'__main__'`.
- If we `import mymod`, then `__name__` will be `'mymod'`'

Who cares? We can use this to distinguish between these two states. We can then have some things happen only if the program is being run standalone. 

This is super common! Nearly every module in Python gives us the opportunity to do so. It does this with a line that is almost universal:

```python
if __name__ == '__main__':
    STUFF
```

This line, which often appears toward the end of a program, basically says, "Anything under this line will only be run if we're *not* imported, and are run directly."

This allows us to have a module that defines variables and functions, but also is executable. Common things done in such an executable:

- Asking the user for input, and then using that input with one or more functions.
- Run automated testing on the module
- Demo what the module can do

Having this line is *NOT* mandatory! It's not at all the same as a `main` function in C.

# Exercise: Using `glob`

1. The `glob` module in the standard library lets you find all of the files that match a particular pattern. Using the `*` and `?` charcters, you can find all of the files you want.
2. Write a program that asks the user to enter a globbing pattern.
3. Find all filenames matching that pattern, using `glob.glob`.
4. For each filename, print the size and the the name. (You can get the size by iterating in a `for` loop over each line, adding `len(one_line)` as you do so.

````python
import glob

pattern = input('Enter a pattern: ').strip()

for one_filename in glob.glob(pattern):
    file_size = 0
    for one_line in open(one_filename):
        file_size += len(one_line)  # add the length of the current line to file_size
    print(f'{one_filename}: {file_size}')
````

# An alternative: `pathlib`

`pathlib` is in the standard library, and allows us to work more easily with filenames and files.

````python
from pathlib import Path

p = Path('.') # current directory

pattern = input('Enter a pattern: ').strip()

for one_filename in p.glob(pattern):
    file_size = 0
    for one_line in open(one_filename):
        file_size += len(one_line)  # add the length of the current line to file_size
    print(f'{one_filename}: {file_size}')
````

# PyPI

Python's standard library is rich and amazing... but also limited:

- A new version only comes out when a new version of Python does
- Authors who contribute to the standard library need to sign over their ownership to the PSF
- Only a small number of modules are so popular that they would even be considered -- at a certain point, the download of Python needs to be a reasonable size

PyPI (Python Package Index) is the community's repository for third-party packages. If you want to share a package with other Python programmers, you can upload your work to PyPI. Similarly, if you want to benefit from work done by other Python programmers, you can download and retrieve from PyPI.

It's at PyPI.org.

# Installing packages

In theory, you should be able to say

    pip install PACKAGENAME

on the command line, and it'll just work. However, sometimes (often?) `pip` isn't installed or avaialble easily. You can instead say

    python -m pip install PACKAGENAME

which tends to work more frequently. 

But the biggest issue isn't with pip, but rather with Python's packages. Only one version of a package can be installed at a time. If you have two projects, and one needs rich 15.0, and another needs rich 16.0, then there isn't any real way to solve the conflict. Only one version can exist on your computer at a time.

One way that Python gets around this is with projects, and every project has its own versions/installations of packages. 

You can do that, but it quickly gets tricky.

Fortunately, we now have `uv`, which is not just super duper fast, but handles a lot of these nuances very well. 

You can create a new project with

    uv init PROJECTNAME

If you're inside of PROJECTNAME, you can say

    uv add PACKAGENAME

It'll be added, but only locally, so it won't interfere.

Inside of a project, you run the program with

    uv run COMMAND
