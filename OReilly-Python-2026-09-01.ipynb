# What are modules?

Twice so far, I've talked to you about the DRY rule -- don't repeat yourself! 

- If the same line of code is used several times in a row, then we can replace those lines with a loop.
- If the same code is used in several places in your program, then you can use a function instead.

What happens if you have the same code in several different programs?

Examples:

- Menuing system
- Login system
- Secret formula that you don't want to have in multiple places

In all of these cases, you can't just define a function and use it, because you have multiple programs.

So the solution in most programming languages is a *library*, a collection of data and functionality that you can use repeatedly, and in multiple programs.

In Python, we call a library a "module" or a "package."

- A module is one Python file that we can "import" into our program and use
- A package is a directory/folder of modules, a collection that we can pick and choose from

We use modules all of the time in Python. But in Python, they do more than that:

- Modules provide us with data and functions that we'll use in multiple programs
- Modules provide us with *namespaces*, like surnames for variables
- Packages allow us to distribute a module (or collection of modules) to other Python programmers, in our own organization or even elsewhere.

# Let's use a module!

If you want to use a module, just use the `import` statement. This loads a module from disk, into memory, and defines a new variable with the same name as the module. For example:

```python
import random
```

# Notice a few things

1. `import` is a statement, not a function. No `()` are needed (or used).
2. The argument we give to `import` is not a string, and not a filename. Rather, it's the name of the variable we want to create. HOWEVER, Python does use this variable name to find the appropriate file. When I say `import random`, Python will look for a file called `random.
