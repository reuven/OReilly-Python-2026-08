# What are modules?

Twice so far, I've talked to you about the DRY rule -- don't repeat yourself! 

- If the same line of code is used several times in a row, then we can replace those lines with a loop.
- If the same code is used in several places in your program, then you can use a function instead.

What happens if you have the same code in several different programs?

Examples:

- Menuing system
- Login system
- Secret formula that you don't want to have in multiple places

In all of these cases, you can't just define a function and use it, because you have multiple programs.

So the solution in most programming languages is a *library*, a collection of data and functionality that you can use repeatedly, and in multiple programs.

In Python, we call a library a "module" or a "package."

- A module is one Python file that we can "import" into our program and use
- A package is a directory/folder of modules, a collection that we can pick and choose from

We use modules all of the time in Python. But in Python, they do more than that:

- Modules provide us with data and functions that we'll use in multiple programs
- Modules provide us with *namespaces*, like surnames for variables
- Packages allow us to distribute a module (or collection of modules) to other Python programmers, in our own organization or even elsewhere.

# Let's use a module!

If you want to use a module, just use the `import` statement. This loads a module from disk, into memory, and defines a new variable with the same name as the module. For example:

# Notice a few things

1. `import` is a statement, not a function. No `()` are needed (or used).
2. The argument we give to `import` is not a string, and not a filename. Rather, it's the name of the variable we want to create. HOWEVER, Python does use this variable name to find the appropriate file. When I say `import random`, Python will look for a file called `random.py`.

After `import` has run, `random` now exists as a defined variable. It is a module object, which means it's basically a container/warehouse for other values and names.

# Now that I've imported `random`, what can I do with it?

All of the names in an imported module are available as attributes. Attributes are names that come after a `.`. If you want the value `x` from `random`, you would say `random.x`. If you want to call the function `thefunc` from `random`, you would say `random.thefunc()`.

This raises a question or two: How can I know what names are available?

1. You can use `dir(random)` to get a list of strings showing what names are available -- but it won't say what is a function and what is data, etc.
2. You can use `help(random)` to get more extensive help -- the docstring for the module and its functions.
3. Best of all (but most annoying sometimes) is `docs.python.org`, and get info from the standard library documentation.

````python
dir(random)
````

# Exercise: Guessing game

1. Generate a random `secret` number by using the `randint` function in the `random` module, choosing one from 0-100.
2. Ask the user, repeatedly, to guess the number:
    - If they guess correctly, say "You got it!" and exit from the program
    - If it's too low, say "Too low!"
    - If it's too high, say "Too high!"
3. If they got it wrong, have them try again
4. Print the number of guesses it took for them to get it right.

````python
import random

secret = random.randint(0, 100)
guess_counter = 0

while True:
    guess_text = input('Guess: ').strip()

    if not guess_text.isdecimal():
        print(f'{guess_text} is not numeric; try again')
        continue
    
    guess = int(guess_text)
    guess_counter += 1
    
    if guess == secret:
        print(f'You got it in {guess_counter} guesses!')
        break
    elif guess < secret:
        print(f'Too low!')
    else:
        print(f'Too high!')
````

# We can get around this with `from .. import`

- This is an alternative syntax for `import`
- You say `from MODULE import NAME` -- you can have `NAME1, NAME2`, etc.
- If you do this, then the module is still loaded in its entirety
- If you do this, then the module is not defined as a variable -- only the names you requested are

````python
from random import randint
````

````python
randint(0, 100)  # we defined this!
````

````python
random.randint(0, 100)  # this will work here, because I already imported random
````

# Two more ways to import

I just told you that I can import Pandas with `import pandas`.

No one does that. Everyone, but everyone, uses an alias for `pandas`, `pd`.

How can you create the alias? With the `as` option to `import`.

````python
import pandas as pd   # this imports pandas.py, but then assigns it to the variable pd, and not to pandas
````

Similarly, you can say

```python
from MODULE import NAME as ALIAS
```

This imports `MODULE` into memory, but only defines one name, `ALIAS`, which is assigned to `MODULE.NAME`.

# The 4 (5?) different versions of `import`

- `import MODNAME`
- `import MODNAME as ALIAS`
- `from MODNAME import NAME`
- `from MODNAME import NAME as ALIAS`

There is a 5th version of this:

`from MODNAME import *`

This means: 

- Load `MODNAME` into memory
- Don't define it as a variable
- Instead, iterate over every name in the module, and define it as a global variable

In other words, if a module defines `x`, `y`, and `z`, we'll have those three variables defined. Not as attributes on the module, but as global variables.

I need to re-import it. Even worse, I cannot use `import` on it, because `import` only works once per module per Python session.

To do that, you'll need to use a special module and function:

````python
from importlib import reload   # importlib comes with Python, and reload lets us reload a module from disk
````

# To create a module

1. Create a file whose name ends with `.py` in a directory in `sys.path`. Easiest is to create it in the same directory as your main program. The module name should be short and a legal Python variable name.
2. In the file, define variables and/or functions.
3. A program can then `import` your module.
4. The program that imports your module has access to its variables and functions via attributes.

# What really happens when we say `import`?

Consider this:

- Before we `import menu`, the `menu.menu` function doesn't exist.
- After we `import menu`, it does.
- Sometime during the import, the function was defined.
- This means that `def menu` inside of `menu.py` executed when we imported the menu.

This is something that's very surprising to a lot of people: When you say `import menu`, you aren't just loading it into memory. Rather, you are executing the entire module, from start to finish.

If a module is very big, very slow, very un-modular, this can take some time.

This means that the module is executed, and functions are only around before `def` executes, and variables are only around because `=` executes.
