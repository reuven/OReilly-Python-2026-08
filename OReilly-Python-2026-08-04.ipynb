# Jupyter and writing Python

If you're going to write Python code, you're going to need (a) the Python language and (b) an editor that knows the syntax of Python.

There's nothing wrong with downloading Python from https://python.org, and then installing an editor (e.g., Visual Studio Code or PyCharm), and then writing your Python code.

However, that's often very difficult for people with little or no Python experience. 

I thus suggest that you not try to install things on your computer. Rather, you can use a 100% in-browser version of Python and an in-browser version of Jupyter.

Jupyter is what I'm typing into right now. This version of Jupyter is running on my computer, not totally in the browser. But you can use Jupyter Lite, which is only the browser.

What is Jupyter?

It gives you a Web-based "notebook" approach to programming:

- You type into "cells"
- You can have documentation (like this) and also code (as we'll see in a bit)
- It runs on a server that isn't your computer (unless you want to), which means less installation and maintenance

Jupyter is very popular, especially among data scientists.

# Quick intro to Jupyter

Everything in Jupyter takes place inside of "cells." A cell can contain documentation (in Markdown format) or code (typically in Python).

When I type, I'm typing into a cell. Actually, it depends on what mode the cell is in:

- If the cell is in **EDIT** mode, then anything I type is put into the cell. Like right now. You can enter edit mode by clicking inside of the cell or by pressing ENTER.
- If the cell is in **COMMAND** mode, then anything I type is taken as a command to Jupyter itself, and isn't displayed. You can enter command mode by clicking to the left of the cell or by pressing ESC.

### Some commands:

- `c` -- copy the current cell
- `x` -- cut the current cell
- `v` -- paste the most recently copied/cut cell
- `a` -- add a new cell **ABOVE** the current one
- `b` -- add a new cell **BELOW** the current one
- `y` -- make a cell good for coding in Python
- `m` -- make a cell good for documentation in Markdown

Always: To tell Jupyter that you're done editing a cell, and want to format the documentation or run the code, you can use shift+Enter.

If you go to https://jupyter.org/try-jupyter/lab/ , that takes you to Jupyter Lite, which is a version of Jupyter that doesn't require any installation at all!

# What is a programming language?

When computers were first invented, each computer solved a single problem. If you wanted to solve a different problem, you needed to build a new computer, or rewire the existing one.

At a certain point, it became clear that this was untenable. People created new computers that were *programmable*, that you could instruct to do different things at different times. Those programs were written using *binary code*, meaning 1s and 0s. 

But it was really tedious to write programs in this way. Programming languages were invented. The idea was that you could write in something resembling a human language, and it would be translated into the 1s and 0s that the computer knows how to deal with.

There are hundreds of thousands of programming languages out there. Each one has a different perspective, syntax, and advantages:

- C and C++: These run super fast, in part because the language is already very "low level," similar in many ways to the binary code used by the computer. However, it's very hard to write with these languages.
- Java and C#: These are high-level languages that use "object-oriented programming," a technique to bundle our code into "objects," which are more concrete than many other systems allow. These are much easier to work with than C and C++. But they are still often confusing and frustrating.
- Python was deliberately invented more than 30 years ago to give everyone the power of a serious programming language but without the very difficult learning curve and esoteric syntax. Python has proven itself over the years to be useful in a wide variety of situations.

Python was never designed to run super fast. The main goal is to have code that you can write easily and then read/debug/maintain easily. Also, the language was designed to be very consistent, so that once you learn something, it stays relevant and important and true throughout your Python career.

Seymour Papert talked about Logo as "low floors and high ceilings." It's easy to get into , and you aren't limited in what you want to create/design/run. 



# Who uses Python?

Many problems don't require the fastest execution. In many ways, the fact that you can read/write/maintain Python faster than other languages more than makes up for its slow speed. Python is perfect for an age in which people are expensive and computers are cheap.

Python is especially popular in the following areas:

- Data science and machine learning
- Analyzing data
- Automated testing
- Web applications
- APIs (servers and clients)
- Internet of things
- Education

Python is everywhere, except when speed is necessary. Also: It's not very popular for mobile apps.
