# What is a programming language?

When computers were first invented, each computer solved a single problem. If you wanted to solve a different problem, you needed to build a new computer, or rewire the existing one.

At a certain point, it became clear that this was untenable. People created new computers that were *programmable*, that you could instruct to do different things at different times. Those programs were written using *binary code*, meaning 1s and 0s. 

But it was really tedious to write programs in this way. Programming languages were invented. The idea was that you could write in something resembling a human language, and it would be translated into the 1s and 0s that the computer knows how to deal with.

There are hundreds of thousands of programming languages out there. Each one has a different perspective, syntax, and advantages:

- C and C++: These run super fast, in part because the language is already very "low level," similar in many ways to the binary code used by the computer. However, it's very hard to write with these languages.
- Java and C#: These are high-level languages that use "object-oriented programming," a technique to bundle our code into "objects," which are more concrete than many other systems allow. These are much easier to work with than C and C++. But they are still often confusing and frustrating.
- Python was deliberately invented more than 30 years ago to give everyone the power of a serious programming language but without the very difficult learning curve and esoteric syntax. Python has proven itself over the years to be useful in a wide variety of situations.

Python was never designed to run super fast. The main goal is to have code that you can write easily and then read/debug/maintain easily. Also, the language was designed to be very consistent, so that once you learn something, it stays relevant and important and true throughout your Python career.

Seymour Papert talked about Logo as "low floors and high ceilings." It's easy to get into , and you aren't limited in what you want to create/design/run.

# Variables and assignment

If you have to type each value you want to display, each time you want to display it, that's going to be a very hard-to-maintain, and annoying-to-write, program.

If you have any experience with another programming language, you might be surprised that I dind't have to *declare* the variable in advance, or say what kind of value it will contain. That's the way it works in Python! You don't declare variables. 

If you assign to a variable that already has a value, then the assignment pushes the old value away, and assigns the new one.

# Why no declarations?

In many programming languages, you cannot just assign any value to any variable. The value has a type (e.g., str or int) and the variable has a type, too. This ensures that you don't make big, stupid mistakes, trying to assign a text value to a numeric variable, or vice versa.

This is both to cut down on mistakes, and because of how the languages work, that they need to know how much space to allocate for the value.

In many languages, you would thus say something like this:

    int x
    x = 5

Python doesn't do this, and doesn't need it, because all variables can refer to all values. That's because we aren't storing the value inside of a box of memory labeled with the variable name. Rather, we're just referring to a value elsewhere in memory. That "reference" doesn't take much memory, and is the same for all values.

Python is known as a "dynamic language" for this reason -- variables don't have types, but values do.

# Conditional execution with `if`

So far, our programs have all assumed that they will run from the start to the end. The entirety of the code was run. 

We don't always want that! If there's functionality in our app that exits the app, we don't always want that to run. We only want it to run under certain conditions. 

`if` changes all that. It allows us to say, "This code should only run if such-and-such is the case." 


````python
first_name = input('Enter your name: ')

if first_name == 'Reuven':
    print('Hi, boss!')
    print('Great to see you again.')
else:
    print('Hello, whoever you are!')
````

# What is going on in the above code?

1. We invoke `input`, ask for the user's first name, and assign it to the variable `first_name`.
2. We use `if`. `if` looks to its right, and needs a `True` or `False` value. Fortunately, to `if`'s right, we have a comparison with `==`, which guarantees that we'll get either `True` or `False`.
3. If the condition is `True`, then the if block (lines 4-5) is run.
4. We also have `else` here.
5. If the `if` condition is `False`, then the `else` block runs.

# `if` syntax

- `if` and then a condition (often, not always, with `==`)
- At the end of the line, we have a `:`
- Following the colon, we have a *BLOCK*. And a block in Python means: The code is indented. It must be indented. Traditionally, we use 4 spaces, but if you're using a good Python editor or notebook, it'll do that automatically when you press Enter after the `:`.
- The indented block can be as long or as short as you want, and can include just about any code.
- If the `if` isn't `True`, then the `else` runs. (Assuming there is one; you don't need it.)

Either the `if` block or the `else` block runs. It cannot be both, and it cannot be neither. One, and only one, will run.

The `else` clause and block are optional. If the `if` condition is `False` and there is no `else`, then the program will just continue silently.

```python
if CONDITION:
    if_was_true_block
else:
    if_was_false_block
```

# What if we enter the same word twice? 

We need to take a third possibility into account!

Here's one option for doing that:

````python
first = input('Enter first word: ')
second = input('Enter second word: ')

if first < second:
    print(first + ' comes before ' + second + '.')
else:
    if second < first:
        print(second + ' comes before '  + first + '.')
    else:
        print(first + ' and ' + second + ' are the same!')
````

````python
# but ... yuck. What if we have 3 possibilities? or 10? or 20? Do we really want massively indented if/else statements?

# instead, we'll use a third type of statement, namely "elif" 

# if we have if/elif/else, then first the "if" comparison is checked for a True value
# if it's False, then we check the second (elif)
# if it's False, then we do the else

# exactly one of these clauses will fire.

first = input('Enter first word: ')
second = input('Enter second word: ')

if first < second:
    print(first + ' comes before ' + second + '.')
elif second < first:
    print(second + ' comes before '  + first + '.')
else:
    print(first + ' and ' + second + ' are the same!')
````

# It's time to talk about f-strings

The above code works, but it's a bit ugly.

Aren't you tired of writing `+` between your strings, anyway?

There is a better way, known as an f-string:

- It's another way to create a string
- It uses `''`, just like a regular string
- We put `f` before the opening quote
- Inside of the string, it's precisely like a regular string *except* that we can have `{}`
- Inside of the `{}`, we can put Python variables and expressions
- Any non-string is turned into a string!

````python
x = 10
y = 20

print(f'x = {x}, y = {y} , and x+y = {x+y}')
````

````python
#  rewritten code from above

first = input('Enter first word: ')
second = input('Enter second word: ')

if first < second:
    print(f'{first} comes before {second}.')
elif second < first:
    print(f'{second} comes before {first}.')
else:
    print(f'You entered {first} and {second}, which look the same to me!')
````

# Conditional execution

`if` is perhaps the most important part of programming. It allows us to say that part of our program will only run if something is the case.

We see this all of the time:

- The shape of the cursor changes depending on what part of the screen we're point at
- The font used on the screen changes with context
- User permissions aren't always the same
- Passwords have to match

What if we want to do something, but only if two things are both `True`? Or if at least one of them is `True`?

For these, we have the special words `and` and `or`. The idea is:

- `and` looks to its left and right, and returns `True` if both sides have a `True` value
- `or` looks to its left and right, and returns `True` if one or both are `True`

````python
x = 10
y = 20

# added space for exaggeration/readability

#      True            True
if   x == 10   and   y == 20:
    print('Great! Both are what you want!')
````

````python
#      True    and      False --- >False
if   x == 10   and   y == 55:
    print('Great! Both are what you want!')
````

````python
# or is similar, but either can be True

#      True    or     False --> True
if   x == 10   or   y == 55:
    print('Great! At least one is what you want!')
````

# Exercise: Pig Latin

Pig Latin is a "secret" children's language. The idea is:

- If an English word starts with a vowel (a, e, i, o, u) then add `way` to the word.
- In other cases, move the first letter to the end, and add `ay`

Some examples:

- `computer` -> `omputercay`
- `papaya` -> `apayapay`
- `elephant` -> `elephantway`
- `apricot` -> `apricotway`

You are to:

- Ask the user to enter a word
- Print the translation into Pig Latin.
