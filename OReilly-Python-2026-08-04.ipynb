# Jupyter and writing Python

If you're going to write Python code, you're going to need (a) the Python language and (b) an editor that knows the syntax of Python.

There's nothing wrong with downloading Python from https://python.org, and then installing an editor (e.g., Visual Studio Code or PyCharm), and then writing your Python code.

However, that's often very difficult for people with little or no Python experience. 

I thus suggest that you not try to install things on your computer. Rather, you can use a 100% in-browser version of Python and an in-browser version of Jupyter.

Jupyter is what I'm typing into right now. This version of Jupyter is running on my computer, not totally in the browser. But you can use Jupyter Lite, which is only the browser.

What is Jupyter?

It gives you a Web-based "notebook" approach to programming:

- You type into "cells"
- You can have documentation (like this) and also code (as we'll see in a bit)
- It runs on a server that isn't your computer (unless you want to), which means less installation and maintenance

Jupyter is very popular, especially among data scientists.
