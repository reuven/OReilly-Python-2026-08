# Dictionaries ("dicts")

So far, we've seen a bunch of different data structures:

- Numbers (both `int` and `float`)
- Text (`str`)
- Containers (`list` and `tuple`)

In strings, lists, and tuples, we have an index. That starts at 0, and then goes up to `len(s) - 1`, where `s` the sequence. The index is used to retrieve values, but there isn't any semantic value to the index. It's just where we happened to store something in our string, list, or tuple.

Dictionaries are kind of like lists, in that we can store anything we want. However, unlike lists, we can use almost anything we want as an index. We can use any numbers we want, but we can also use any strings we want. Suddenly, when you can do that, you discover that there is a very large number of problems that are easier to solve with dicts rather than with lists.

Dicts are very convenient for this reason. But it turns out that they're also extremely efficient. If you know the key (i.e., the index) for a particular value, then you can retrieve it almost instantly, without having to iterate over all of the other keys first. 

# Dicts aren't unique to Python

Other languages have the equivalent to dicts:

- Hash tables
- Hash maps
- Maps
- Hashes
- Key-value stores
- Name-value stores
- Associative arrays

In all cases, regardless of what you call them, dicts are always stored and retrieved as pairs -- key-value pairs. You cannot talk about a dict's individual items; it's always key-value pairs (which Python often calls one "item").

You use the key to get the value. You cannot use the value to get the key!

# Rules for dicts

1. A dict can contain any number of key-value pairs. The minimum is zero, and the maximum is determined by your computer's memory.
2. Any immutable value in Python (e.g., numbers and strings) can be a key, so long as it doesn't contain a mutable value. Normally, we use ints and strings as dict keys.
3. Any value at all is allowed in a dict. (The keys have restrictions, but the values don't.)
4. In a given dict, the keys must be unique.
5. Every key has a value, and every value has a key.

# Terminology

- Dictionary: The data structure in Python that we use for key-value pairs. Other languages call these other things, but in Python, they're always "dict" or "dictionary." A dict contains any number of key-value pairs.
- Hashing: A "hash function" is the special function that helps Python to place the key-value pairs in memory. The reason why other languages call dictionaries "hashes" or "hash maps" is that this hash function is extremely important. When we store a key-value pair, Python runs the hash function on our behalf. When we search for or retrieve a key-value pair, the hash function is again employed.
- Mutable: A data structure can change. In Python, lists and dictionaries are mutable -- we can change them, by adding, replacing, and deleting values.
- Immutable: The opposite of mutable. The data structure cannot change. Once defined, it stays forever. In Python, this includes integers, floats, strings, and tuples. You can get a new value back from an immutable data structure, and then use that instead.
-
- Only immutable values can be used as keys in dicts. For example, ints and strings are immutable, and they're classic keys in dicts.
- If you try to use a mutable value, such as a list, as a dict key, you'll get an error -- indicating that it's "unhashable."

# Paradigms 1 and 2

There are certain patterns of dict use that repeat across many programs. I've identified three that I see all of the time, which I call paradigms 1, 2, and 3.

1. We create a dict at the start of the program, and never change it. We treat it like a read-only database in our program.
2. We create a dict at the start of the program, with keys and default values. We won't add new keys to the dict, and we won't remove existing keys from it. But we will update the values. This is perfect for a situation where we know what we want to count or accumulate, and just want to use the dict for it.

The values in a dict can be integers (as we've seen), but they can also be strings, and they can be other things, too -- such as lists or dicts. A dict containing list values can be a fairly complex data structure. A dict containing dicts as values is what we call a "tree" in programming.

# Exercise: Vowels, digits, and others (dict-list edition)

1. Define a dict, `counts`, with three keys (`vowels`, `digits`, and `others`) and values. All three values should start with an empty list, `[]`.
2. Ask the user to enter a string.
3. Go through the string, one character at a time.
    - If it's a vowel, append `one_character` to the `vowels` list
    - If it's a digit, append `one_character` to the `digits` list
    - Otherwise, append `one_character` to the `others` list

# Paradigm 3: Counting the unknown 

So far, in both dict paradigms, we've known what we want to keep track of. In paradigm 1, it's all set and isn't changing. In paradigm 2, the keys are set from the beginning.

But what if we don't know what keys we'll get? What if we don't know how many we'll get? But .. what if we do know what we want to do once we get the key?

The way that paradigm 3 works is: 

- Start with a completely empty dict
- When we get a key-value pair (or sometimes just a key), we check to see if it's already in the dict
    - If so, we just have to update the value, because the key is already there
    - If not, then we add both the key (which is new) and the value (which is its starting point)

# Example: Counting characters

Let's say we want to know how often each character appears in user input. I could use a dict for this, where the keys are the characters and the values are the number of times each appears.

Maybe I should use paradigm 2, and pre-load the dict with all of the possible characters as keys. That is a truly bad idea -- because there are so many characters, from all languages, plus emojis, that we'll spend enormous amounts of memory and time putting together a dict that is overkill.

Instead, we can use paradigm 3:

- Start with an empty dict
- As we get a new character, check if it's already in the dict
    - If not, then we add a new key (that character) and 1 (the initial count)
    - If so, then we just add 1 to the key's existing value

```python
counts = {}

text = input('Enter text: ').strip()

for one_character in text:
    if one_character in counts:      # if one_character is a key in counts...
        counts[one_character] += 1   # ... add 1 to the existing value
    else:
        counts[one_character] = 1    # add a new key-value pair, starting the count at 1

print(counts)        
```

```python
rainfall = {}

while True:
    city_name = input('City: ').strip()

    if city_name == '':
        break
    
    mm_rain = input('Rain: ').strip()
    mm_rain = int(mm_rain)
    
    if city_name in rainfall:            # if we've seen this city before,
        rainfall[city_name] += mm_rain   #    add its rainfall to the dict for the city
    else:
        rainfall[city_name] = mm_rain    # otherwise, add a new key-value pair

print(rainfall)
```

# What happens if we iterate?

We know that a `for` loop gets different results from different data structures:

- On a string, we get the characters
- On a range, we get integers
- On a list or a tuple, we get the elements

What happens if we iterate over a dict?

```python
d = {'a':10, 'b':20, 'c':30}

for one_thing in d:
    print(one_thing)
```

Iterating over a dict gives us the keys. Not the keys + values, just the keys. But of course, we can use the keys to get the values, too.

```python
for one_key in d:
    print(f'{one_key}: {d[one_key]}')
```

You might see some people invoking a `dict.keys` method on dictionaries, and then either iterating over that or using `in` on the keys. This isn't wrong, in that you'll get the right answer. But it's not considered the best style -- at best, it's just longer and more complex. And at worst, it's actually a little bit slower.

```python
# wouldn't it be nice to get the key and value together?
# we can get that with the "items" method. It returns, with each iteration, a 2-element tuple of (key, value)

for t in d.items():
    print(t)
```

```python
for t in d.items():
    key, value = t   # unpacking!
    print(f'{key}: {value}')
```

```python
# we can do better -- this is my favorite way to iterate over a dict

for key, value in d.items():
    print(f'{key}: {value}')
```

# How do dicts work?

It's easy to think that dicts are just fancy lists. But they're implemented in a completely different way.

When you store a new element in a list, it just gets tacked onto the end, and the index for that element is 1 more than the maximum index was before.

When you store a new key-value pair in a dict, you are dictating the key as well as the value. So it isn't just stored at the end of everything else. Rather, it's stored in a particular place. The method for storage is key to the advantages of a dict.

- When you store `d['a'] = 10` in a dict, Python takes the key (`'a'` in this case) and runs it through a "hash function." In Python, it's available as `hash`, so you can run `hash('a')`. It'll give you what looks like a random number. But it isn't! It's very big, it seems random, and it's only stable within one run of Python -- so if you run your program a second time, it'll give totally different numbers. The big deal is that you cannot easily predict what the output will be for `hash('a')`. That allows Python to choose a (seemingly) random place to store the key and value each time.
- When you retrieve `d['a']`, Python runs `hash('a')` again and jumps immediately to the location in memory where it belives your key-value pair is stored.
- This means that searching for a key in a dict has nothing to do with the size of the dict. You can have 10 pairs or 10m pairs, and it'll still take the same (quick) time to find out whether a key is in the dict. So searching is super fast.
- Retrieving is basically a bit more work than searching, and is thus also very fast.

This all explains a lot:

- How dicts are so fast
- How dicts can use anything (just about) as a key
- Why keys must be unique within a dict

It doesn't explain why we need our keys to be immutable. What's wrong with having a list as a key?

The problem is basically that `hash(my_key)` will be run not only when storing, but also when retrieving. What happens if `my_key` has changed? `hash(my_key)` might thus change, too. That means we won't be able to retrieve the value for that key, because it'll get "lost" in the dict.

Python expects us not to change the keys, and thus basically requires keys to be immutable.

# Files

We use files all of the time: PDF files, Excel files, Word files... but what is a file?

At the end of the day, a file is a way that we can permanently store data structures from our program. 

When we "save" from a program, it's taking all of the data structures in memory, translating them into a way that can be stored, and then stores them in that file.

When we "load" from a program, we're taking those bits on disk, and turning them back into data structures.

We thus have many "binary" file formats, such as Excel, PDF, etc., which are very efficient. But we can also use text files, with just plain letters, numbers, and symbols, to store our data, and then to retrieve it. It's just not going to be as fast or sophisticated.

When we talk about files, we're mostly talking in this class about text files -- either that we'll use to store data structures simply, or that actually contain just text.



# How can we use a file?

We have to ask the OS to help us. We cannot talk to disks ourselves, directly. We must get a go-between assigned to us by the OS.

- We ask the OS to open a file
- It either doesn't (for whatever reason) or does, and gives us a "file handle" or "file object" in response.
- This file handle lets us work with the file -- normally just reading, but we'll see about writing later.
- When we're done, we close the file, thus returning the resources to the OS.

At no point are we talking to the file directly! The OS is involved at every step.

# Technically speaking, how does it work?

- The `open` function takes an argument, a string, which is a filename. By default, it allows us to read from the file, using the `read` method. (We'll see some others soon.)
- When you're done with a file, you should close it, by invoking the `close` method on the object you got.

# Exercise: IP addresses

In the zipfile I gave to you is a file called `mini-access-log.txt`. It contains information about every request to a server I ran 15+ years ago. At the start of each line is an IP address. 

Our goal is to find out how many requests came from each IP address in the logs.

0. Define `counts`, an empty dict
1. Open the file
2. Iterate over it, one line at a time
3. Grab the IP address from the start of each line
4. Either add a new key-value pair for the IP address (if it's new)
5.  ... or add 1 to counts for this IP address
6. Iterate over `counts` and print each IP address and its number.

```python
counts = {}

f = open('mini-access-log.txt')

for one_line in f:
    fields = one_line.split()   # split on any whitespace, of any length, and any combination
    ip_address = fields[0]

    if ip_address in counts:       # if we've seen it before, then...
        counts[ip_address] += 1    # add 1 to the count for this IP address
    else:
        counts[ip_address] = 1     # otherwise, add a new key-value pair for the IP address

f.close()

for key, value in counts.items():
    print(f'{key}\t{value}')
```

# Exercise: User dict

The `passwd.txt` file that we saw before has a number of fields, including the username (field index 0) and the user ID number (which is field index 2). 

1. Create an empty dict called `users`.
2. Iterate over `passwd.txt`, one line at a time. Grab the username and the user ID from each line.
3. Create a new key-value pair in the `users` dict for that user.
4. The end result will be a dictionary that reflects the contents of the file, at least so far as username and user ID are concerned.

```python
# create a dict with usernames as keys and user IDs as values

users = {}

f = open('passwd.txt')

for one_line in f:
    fields = one_line.split(':')
    username = fields[0]
    user_id = int(fields[2])   # user ID is an integer

    users[username] = user_id

print(users)    
```
