# What are functions?

If you think of the data in a programming language as the nouns, then the functions are the verbs. 

We have already seen a number of functions:

- `print`
- `input`
- `len`
- `sum`
- `type`

In each case, we invoke the function with `()`. That tells Python to run the function and to give us whatever it returned.

Why, though, would we want functions?

1. The DRY (don't repeat yourself) rule -- if you have similar functionality in many places in your program, then it's best to write it once, defining it as a function, and then invoke that function whenever you need. If you ever need to update the function, it's in one place, not copied over several different places.
2. Having a function in one place also lowers the cognitive load -- you don't need to remember as much code, because there isn't as much code!
3. Abstraction -- ignore the details, and focus on the big idea. By defining a function, I wrap up a lot of actions under one name, and then I can think about that one name, rather than all of the actions. This makes it easier to think about, but also easier to build larger systems, and also easier to communicate with other people about it.

Functions are, in many ways, the main building block of programming. 

What functions you have to write depends on what type of work you're doing. It also depends on the data structure you're using. Choosing data structures wisely can make your functions easier to write, and also more efficient.

# Return values

It's a good idea for functions to return values, not to print them. If a function prints, then the value it printed is lost. But if it returns a value, then the caller has the option of printing it, storing it, calculating with it, etc.

So returning is always more flexible.

In addition to returning a value, `return` means: Stop execution of the function here.

You can return a value with the `return` statement:

- If you don't explicitly `return` from a function, it returns `None`.
- If you just say `return` in a function, it also returns `None`
- Otherwise, you can return any value you want.
- `return` isn't a function, so we typically don't put the returned value in `()`.

```python
def hello(first_name):
    print(f'Hello, {first_name}!')

s = hello('Reuven')    
```

```python
print(s)
```

```python
# now let's change it to return a value

def hello(first_name):
    return f'Hello, {first_name}!'

s = hello('Reuven')      # value was "captured" by s
```

```python
print(s)
```

- Can a function both print and return? Yes! But it's still normally a bad idea to print from a function.
- Can a function have more than one `return` statement? Yes, but the first one executed ends the function's run.
- What can we return from a function? Any Python value at all -- integer, string, list, tuple, dict, etc etc etc.

- How can we stop people from invoking our function with the wrong type of argument?
- How can we advertise what the "right" type of argument is?

Because Python is a "dynamically typed" language, any variable can refer to any value. Which means that there isn't anything checking what type of value we're passing to a function when we call it.

Python does support "type hints," but the language itself doesn't enforce them, and you need external tools to pay attention. 

Traditionally, Python has used "docstrings" to document what arguments a function expects to get. Still today, every major function in Python (especially those that come with the language) use docstrings to indicate what values are acceptable. Moreover, a docstring can tell someone what a function does, what it changes, and what it returns.

### What is a docstring?

It's a string... containing documentation.

To be treated specially, it must be on the first line of a function. And the string isn't assigned to a variable. Rather, it's just placed there, as a string, on that first line.

Traditonally, we use triple-quoted strings for docstrings, just to have more than one line available.

```python
def analyze_word(text):
    counts = {}

    for one_character in text:
        if one_character in counts:    # if one_character is a key in the "counts" dict
            counts[one_character] += 1 # ... add 1 to its value
        else:
            counts[one_character] = 1  # otherwise, create a new key-value pair
    
    return text, len(text), counts
```

# Keyword arguments

Until now, all of the arguments we've passed to a function have been *positional* arguments. Meaning: The first argument is assigned to the first parameter. The second to the second, etc etc

Keyword arguments allow us to tell the function which values should be assigned to which parameters. Or, even if it would be the same order as positional arguments, often it's clearer to use keyword arguments.

These look a bit different from positional arguments, in that kw args are all of the form `name=value`.

```python
def sub(x, y):
    return x - y

```

```python
# parameters: x   y
# arguments: 10   5

sub(10, 5)
```

```python
# parameters:  x    y
# arguments:   10  5

sub(x=10, y=5)
```

```python
# parameters:  x    y
# arguments:   10  5

sub(y=5, x=10)
```

# Rules for keyword arguments

1. They always look like `name=value`.
2. The `name` will be turned into a variable name, and has to match one of the parameters.
3. The most important thing: All positional arguments must come before all keyword arguments.

```python
sub(5, y=10)
```

```python
sub(x=5, 10)   # this won't go well!
```

```python
sub(5, x=10)   # this won't go well!
```

# Exercise: Count characters

1. Define a function, `count_characters`, that takes two arguments:
    - `text`, a string
    - `to_count`, another string
2. The function will return an integer, the number of times that any character in `to_count` appears in `text`.
3. If we invoke the function with `to_count='aeiou'`, then we're counting the vowels. If we invoke it with `to_count='0123456789'`, we're counting the digits.
4. Once the function works, invoke it three times:

```python
def count_characters(text, to_count):
    # how many characters in text appear in to_count?
    output = 0

    for one_character in text:          # go through each character in text
        if one_character in to_count:   # if the current character is countable
            output += 1

    return output
```

```python
def count_characters(text, to_count='aeiou', lowercase=False):
    count = 0

    if lowercase:
        text = text.lower()

    for one_character in text:
        if one_character in to_count:
            count += 1

    return count

count_characters('hello out there', 'aeiou')    
```

# Exercise: `low_high`

3. The function will return a 2-element list with the lowest and then highest integers from the input arguments.
4. Yes, the low and high can be the same.

    low_high(10, 20, 5, 30, 15) # returns [5, 30]

```python
def low_high(numbers):
    lowest = numbers[0]
    highest = numbers[0]

    for one_number in numbers:
        if one_number < lowest:   # lowest than the existing lowest?
            lowest = one_number   # assign!

        if one_number > highest:  # highest than the existing highest?
            highest = one_number  # assign!

    return [lowest, highest]
```

```python
def low_high(*numbers):   # now I'm saying: Don't pass me a list, but just the numbers themselves
    lowest = numbers[0]
    highest = numbers[0]

    for one_number in numbers:
        if one_number < lowest:   # lowest than the existing lowest?
            lowest = one_number   # assign!

        if one_number > highest:  # highest than the existing highest?
            highest = one_number  # assign!

    return [lowest, highest]
```

# Local vs. global variables

What happens if two different functions have the parameter `x`? Can they call one another. Does one's `x` get priority?

In a function, all of the parameters are local variables. Local variables are not available outside of a function. Once the function ends, then local variables all disappear, and are not seen.

This means that your function is basically isolated from other functions. That is a good thing! It means you can write your funciton with whatever names you want, and it won't affect any other function, or be affected them, either.

# Warning about defaults

Because Python is compiled into opcodes, and then executed, there are little things in Python that can end up biting you.

One of them is default argument values. If you use mutable defaults, you will pay for it! You really want to make sure that any default is immutable -- number, string, or (maybe even) tuple. But using a list or dict for your default has very bad potential impliciations.

Every editor will warn you about "dangerous mutable defaults." Pay attention to this warning!
