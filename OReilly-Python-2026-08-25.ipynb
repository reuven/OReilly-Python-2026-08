# What are functions?

If you think of the data in a programming language as the nouns, then the functions are the verbs. 

We have already seen a number of functions:

- `print`
- `input`
- `len`
- `sum`
- `type`

In each case, we invoke the function with `()`. That tells Python to run the function and to give us whatever it returned.

Why, though, would we want functions?

1. The DRY (don't repeat yourself) rule -- if you have similar functionality in many places in your program, then it's best to write it once, defining it as a function, and then invoke that function whenever you need. If you ever need to update the function, it's in one place, not copied over several different places.
2. Having a function in one place also lowers the cognitive load -- you don't need to remember as much code, because there isn't as much code!
3. Abstraction -- ignore the details, and focus on the big idea. By defining a function, I wrap up a lot of actions under one name, and then I can think about that one name, rather than all of the actions. This makes it easier to think about, but also easier to build larger systems, and also easier to communicate with other people about it.

Functions are, in many ways, the main building block of programming. 

What functions you have to write depends on what type of work you're doing. It also depends on the data structure you're using. Choosing data structures wisely can make your functions easier to write, and also more efficient.

# How do we define a function?

1. We use the keyword `def`.
2. After that (on the same line), we name the function. This is a variable name!
3. After the function name (for now), we'll put `()`. We will soon fill these with parameter names, but not just yet.
4. At the end of the line, we have `:`
5. On the following line, we have the *function body*. This is an indented block that can contain any Python code we want.

```python
def hello():
    print('Hello!')
```

```python
# we just defined the function. When you define it, the function doesn't run.
# You have to invoke the function for that.

hello()  
```

```python
# what if I want to invoke the function a number of times?

for i in range(3):
    hello()
```

```python
# let's try another version of this function

def hello():
    first_name = input('Enter your name: ').strip()
    print(f'Hello, {first_name}!')
```

# A few things to notice

1. I just defined a new function named `hello`. My previous function was named ... `hello`. I have now replaced the old definition with the new one. Only one function with a given name can exist in Python at a time.
2. A function's body isn't executed when we define the function, just when we invoke it.
3. The function body can contain any code we want. Here, we are invoking `input`, assigning to a variable, and invoking `print`.

```python
hello()
```

```python
hello  # no parentheses -- it shows me the function object, which can be called... but it isn't
```

# Exercise: Calculator

1. Define a function, `calc`, that when invoked asks three questions:
    - What is the first number?
    - What is the operation?
    - What is the second number?
2. The answer to each question should be stored to a separate variable.
3. If the numbers can be turned into integers, and if the operator is known (`+` or `-`), then get the result of the calculation and print it, along with the numbers and the operator.
4. If the number inputs aren't really digits, then scold the user.
5. If the operator is not known, then the result can be `(no result)`.

Example:

    calc()
    First: 10
    Operator: +
    Second: 30
    10 + 30 = 40

    calc()
    First: 10
    Operator: +
    Second: hello
    hello is not numeric

    calc()
    First: 10
    Operator: *
    Second: 30
    10 * 30 = (no result)

    

    

```python
def calc():
    first = input('First: ').strip()
    op = input('Operator: ').strip()
    second = input('Second: ').strip()

    if op == '+':
        result = first + second
    elif op == '-':
        result = first - second
    else:
        result = '(Not supported)

    print(f'first = {first}, op = {op}, second = {second}')
```
