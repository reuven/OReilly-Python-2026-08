# What are functions?

If you think of the data in a programming language as the nouns, then the functions are the verbs. 

We have already seen a number of functions:

- `print`
- `input`
- `len`
- `sum`
- `type`

In each case, we invoke the function with `()`. That tells Python to run the function and to give us whatever it returned.

Why, though, would we want functions?

1. The DRY (don't repeat yourself) rule -- if you have similar functionality in many places in your program, then it's best to write it once, defining it as a function, and then invoke that function whenever you need. If you ever need to update the function, it's in one place, not copied over several different places.
2. Having a function in one place also lowers the cognitive load -- you don't need to remember as much code, because there isn't as much code!
3. Abstraction -- ignore the details, and focus on the big idea. By defining a function, I wrap up a lot of actions under one name, and then I can think about that one name, rather than all of the actions. This makes it easier to think about, but also easier to build larger systems, and also easier to communicate with other people about it.

Functions are, in many ways, the main building block of programming. 

What functions you have to write depends on what type of work you're doing. It also depends on the data structure you're using. Choosing data structures wisely can make your functions easier to write, and also more efficient.

# How do we define a function?

1. We use the keyword `def`.
2. After that (on the same line), we name the function. This is a variable name!
3. After the function name (for now), we'll put `()`. We will soon fill these with parameter names, but not just yet.
4. At the end of the line, we have `:`
5. On the following line, we have the *function body*. This is an indented block that can contain any Python code we want.

```python
def hello():
    print('Hello!')
```

```python
# we just defined the function. When you define it, the function doesn't run.
# You have to invoke the function for that.

hello()  
```

```python
# what if I want to invoke the function a number of times?

for i in range(3):
    hello()
```

```python
# let's try another version of this function

def hello():
    first_name = input('Enter your name: ').strip()
    print(f'Hello, {first_name}!')
```

# Exercise: Calculator

1. Define a function, `calc`, that when invoked asks three questions:
    - What is the first number?
    - What is the operation?
    - What is the second number?
2. The answer to each question should be stored to a separate variable.
3. If the numbers can be turned into integers, and if the operator is known (`+` or `-`), then get the result of the calculation and print it, along with the numbers and the operator.
4. If the number inputs aren't really digits, then scold the user.
5. If the operator is not known, then the result can be `(no result)`.

Example:

    calc()
    First: 10
    Operator: +
    Second: 30
    10 + 30 = 40

    calc()
    First: 10
    Operator: +
    Second: hello
    hello is not numeric

    calc()
    First: 10
    Operator: *
    Second: 30
    10 * 30 = (no result)

    

    

```python
def calc():
    first = input('First: ').strip()
    op = input('Operator: ').strip()
    second = input('Second: ').strip()

    if first.isdigit() and second.isdigit():

        first = int(first)
        second = int(second)
    
        if op == '+':
            result = first + second
        elif op == '-':
            result = first - second
        else:
            result = '(Not supported)'
    
        print(f'{first} {op} {second} = {result}')

    else:
        print(f'{first} and {second} both need to be numbers; try again')
```

# Loop refresher

- `for` loops go through each element of an "iterable":
    - Iterating over `range(n)` gives us integers from 0 until (not including) `n`
    - Iterating over a string gives us its characters
    - Iterating over a list or tuple gives us its elements
    - Iterating over a dict gives us its keys
    - Iterating over `dict.items()` gives us `(key, value)` 2-element tuples
    - Iterating over a file gives us each line of the file as strings, each ending with `\n`
- `while` loops are like `if`, with a condition
    - If the condition is `True`, the loop body runs, and then returns to the `while` and condition for a potential additional run


```python
def calc():
    first = int(input('what is the first number:' ))
    operator = input('what is the operation:' )
    second = int(input('what is the second number: ' ))	 

    if operator == '+':
        print (first+second)
    elif operator == '-':
        print (first - second)
    elif operator == '*':
        print (first * second)						 
    else:
        print(f'Bad operator {operator}')
```

```python
calc()
```

# Arguments and parameters

- When you invoke a function, you can pass one or more arguments to it. These are values.
- The function itself, in order to accept the arguments, needs to have *parameters*, variables to which arguments are assigned.

The difference between parameters and other variables is that you normally need to assign to a variable with `=`. But parameters get their values via the function call.

```python
def hello(first_name):   # now my function has one parameter -- so we'll need to get one argument
    print(f'Hello, {first_name}!')
```

# Return values

It's a good idea for functions to return values, not to print them. If a function prints, then the value it printed is lost. But if it returns a value, then the caller has the option of printing it, storing it, calculating with it, etc.

So returning is always more flexible.

In addition to returning a value, `return` means: Stop execution of the function here.

You can return a value with the `return` statement:

- If you don't explicitly `return` from a function, it returns `None`.
- If you just say `return` in a function, it also returns `None`
- Otherwise, you can return any value you want.
- `return` isn't a function, so we typically don't put the returned value in `()`.

```python
def hello(first_name):
    print(f'Hello, {first_name}!')

s = hello('Reuven')    
```

```python
print(s)
```

```python
# now let's change it to return a value

def hello(first_name):
    return f'Hello, {first_name}!'

s = hello('Reuven')      # value was "captured" by s
```

```python
print(s)
```

- Can a function both print and return? Yes! But it's still normally a bad idea to print from a function.
- Can a function have more than one `return` statement? Yes, but the first one executed ends the function's run.
- What can we return from a function? Any Python value at all -- integer, string, list, tuple, dict, etc etc etc.

- How can we stop people from invoking our function with the wrong type of argument?
- How can we advertise what the "right" type of argument is?

Because Python is a "dynamically typed" language, any variable can refer to any value. Which means that there isn't anything checking what type of value we're passing to a function when we call it.

Python does support "type hints," but the language itself doesn't enforce them, and you need external tools to pay attention. 

Traditionally, Python has used "docstrings" to document what arguments a function expects to get. Still today, every major function in Python (especially those that come with the language) use docstrings to indicate what values are acceptable. Moreover, a docstring can tell someone what a function does, what it changes, and what it returns.

### What is a docstring?

It's a string... containing documentation.

To be treated specially, it must be on the first line of a function. And the string isn't assigned to a variable. Rather, it's just placed there, as a string, on that first line.

Traditonally, we use triple-quoted strings for docstrings, just to have more than one line available.

```python
def hello(first_name):
    return f'Hello, {first_name}!'
```

```python
hello('Reuven')
```

```python
hello(5)
```

```python
# let's add a docstring!

def hello(first_name):
    """A friendly function that greets someone.

    Expects: A string argument, presumably containing someone's name.
    Modifies: Nothing
    Returns: A string containing a friendly greeting and the person's name.
    """
    return f'Hello, {first_name}!'
```

```python
hello('Reuven')
```

```python
# if I use the "help" function to get information about this "hello" function...

help(hello)  # I'm not invoking hello! I'm just giving it as an argument to help!
```

# How are docstrings different from comments?

- Comments are meant for the people who will maintain a function. They need to know how things are implemented.
- Docstrings are meant for the people who will invoke a function. They need to know what inputs and outputs are expected.

These are different populations, albeit with some obvious overlap.

# Exercise: `analyze_word`

1. Write a function, `analyze_word`, which takes one string as an argument
2. The function returns a 3-element tuple:
    - the original word (a string)
    - the length of the word (an integer)
    - each character in the string, as keys in a dict -- the values represent how many times each character appeared
3. invoke the function, and use unpacking to retrieve all three elements.

Example:

    analyze_word('hello')

    ('hello', 5, {'h':1, 'e':1, 'l':2, 'o':1})

```python
def analyze_word(text):
    counts = {}

    for one_character in text:
        if one_character in counts:    # if one_character is a key in the "counts" dict
            counts[one_character] += 1 # ... add 1 to its value
        else:
            counts[one_character] = 1  # otherwise, create a new key-value pair
    
    return text, len(text), counts
```

```python
analyze_word('hello')
```
